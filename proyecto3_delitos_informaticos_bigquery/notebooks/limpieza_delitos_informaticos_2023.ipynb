```python
import pandas as pd

ruta_archivo = "Compendios_especialidades_2023.xlsx"

df_original = pd.read_excel(
    ruta_archivo,
    sheet_name="CA_DelitosInf",
    header=None
)

print("Dimensiones originales:", df_original.shape)

display(df_original.iloc[:25, :15])
```

```python
pd.set_option("display.max_columns", 20)
pd.set_option("display.max_colwidth", 60)

display(df_original.iloc[13:29, 0:13])
```

```python
columnas_andalucia = [
    "categoria_delito",
    "tipo_delito",
    "indicador",
    "Almería",
    "Cádiz",
    "Córdoba",
    "Granada",
    "Huelva",
    "Jaén",
    "Málaga",
    "Sevilla",
    "ANDALUCÍA"
]

df_andalucia = df_original.iloc[20:88, 0:12].copy()
df_andalucia.columns = columnas_andalucia

display(df_andalucia.head(12))
```

```python
df_andalucia[["categoria_delito", "tipo_delito"]] = (
    df_andalucia[["categoria_delito", "tipo_delito"]].ffill()
)

display(df_andalucia.head(12))
```

```python
df_andalucia_largo = df_andalucia.melt(
    id_vars=[
        "categoria_delito",
        "tipo_delito",
        "indicador"
    ],
    var_name="territorio",
    value_name="valor"
)

print("Dimensiones:", df_andalucia_largo.shape)

display(df_andalucia_largo.head(15))
```

```python
df_andalucia_largo["anio"] = 2023
df_andalucia_largo["comunidad_autonoma"] = "Andalucía"
df_andalucia_largo["nivel_territorial"] = "provincia"

df_andalucia_largo.loc[
    df_andalucia_largo["territorio"] == "ANDALUCÍA",
    "nivel_territorial"
] = "comunidad_autonoma"

df_andalucia_largo = df_andalucia_largo[
    [
        "anio",
        "comunidad_autonoma",
        "nivel_territorial",
        "territorio",
        "categoria_delito",
        "tipo_delito",
        "indicador",
        "valor"
    ]
]

display(
    pd.concat([
        df_andalucia_largo.head(3),
        df_andalucia_largo.tail(3)
    ])
)
```

```python
print("Valores nulos por columna:")

display(
    df_andalucia_largo
    .isna()
    .sum()
    .to_frame(name="cantidad_nulos")
)

print("Tipos de datos:")

display(
    df_andalucia_largo
    .dtypes
    .to_frame(name="tipo_dato")
)
```

```python
nulos_por_territorio = (
    df_andalucia_largo[
        df_andalucia_largo["valor"].isna()
    ]
    .groupby("territorio")
    .size()
    .reset_index(name="cantidad_nulos")
)

display(nulos_por_territorio)
```

```python
df_andalucia_largo["valor"] = (
    df_andalucia_largo["valor"].astype("Int64")
)

print("Tipo de dato de valor:", df_andalucia_largo["valor"].dtype)
print("Valores nulos conservados:", df_andalucia_largo["valor"].isna().sum())

display(
    df_andalucia_largo.loc[
        df_andalucia_largo["valor"].isna(),
        ["territorio", "tipo_delito", "indicador", "valor"]
    ].head(8)
)
```

```python
claves = [
    "categoria_delito",
    "tipo_delito",
    "indicador"
]

suma_provincias = (
    df_andalucia_largo[
        df_andalucia_largo["nivel_territorial"] == "provincia"
    ]
    .groupby(claves, as_index=False)["valor"]
    .sum(min_count=1)
    .rename(columns={"valor": "suma_provincias"})
)

totales_andalucia = (
    df_andalucia_largo[
        df_andalucia_largo["nivel_territorial"] == "comunidad_autonoma"
    ][claves + ["valor"]]
    .rename(columns={"valor": "total_andalucia"})
)

control_totales = suma_provincias.merge(
    totales_andalucia,
    on=claves,
    how="outer",
    validate="one_to_one"
)

control_totales["diferencia"] = (
    control_totales["suma_provincias"]
    - control_totales["total_andalucia"]
)

display(
    control_totales["diferencia"]
    .value_counts(dropna=False)
    .rename_axis("diferencia")
    .reset_index(name="cantidad")
)
```

```python
inicios_bloques = [
    numero_columna
    for numero_columna, contenido in enumerate(df_original.iloc[13])
    if isinstance(contenido, str)
    and contenido.strip().upper().startswith("FISCAL")
]

bloques_detectados = pd.DataFrame({
    "columna_inicio": inicios_bloques,
    "encabezado_original": [
        df_original.iloc[13, columna].strip()
        for columna in inicios_bloques
    ]
})

print("Comunidades autónomas detectadas:", len(bloques_detectados))

display(bloques_detectados)
```

```python
columna_total_nacional = next(
    numero_columna
    for numero_columna, contenido in enumerate(df_original.iloc[13])
    if isinstance(contenido, str)
    and contenido.strip().upper() == "TOTAL NIVEL NACIONAL"
)

limites_bloques = (
    inicios_bloques[1:]
    + [columna_total_nacional]
)

resumen_bloques = []

for inicio, limite in zip(inicios_bloques, limites_bloques):
    territorios = [
        contenido.strip()
        for contenido in df_original.iloc[19, inicio:limite]
        if isinstance(contenido, str)
        and contenido.strip()
    ]

    resumen_bloques.append({
        "columna_inicio": inicio,
        "numero_territorios": len(territorios),
        "territorios_detectados": territorios,
        "total_comunidad": territorios[-1]
    })

df_resumen_bloques = pd.DataFrame(resumen_bloques)

display(df_resumen_bloques)
```

```python
bloques_largos = []

for inicio, limite in zip(inicios_bloques, limites_bloques):

    columnas_territorio = [
        columna
        for columna in range(inicio + 3, limite)
        if isinstance(df_original.iloc[19, columna], str)
        and df_original.iloc[19, columna].strip()
    ]

    territorios = [
        df_original.iloc[19, columna].strip()
        for columna in columnas_territorio
    ]

    columnas_bloque = [
        inicio,
        inicio + 1,
        inicio + 2
    ] + columnas_territorio

    df_bloque = df_original.iloc[
        20:88,
        columnas_bloque
    ].copy()

    df_bloque.columns = [
        "categoria_delito",
        "tipo_delito",
        "indicador"
    ] + territorios

    df_bloque[
        ["categoria_delito", "tipo_delito"]
    ] = df_bloque[
        ["categoria_delito", "tipo_delito"]
    ].ffill()

    df_bloque_largo = df_bloque.melt(
        id_vars=[
            "categoria_delito",
            "tipo_delito",
            "indicador"
        ],
        var_name="territorio",
        value_name="valor"
    )

    total_comunidad = territorios[-1]

    nombre_comunidad = (
        total_comunidad
        .title()
        .replace(" Y ", " y ")
    )

    df_bloque_largo["anio"] = 2023
    df_bloque_largo["comunidad_autonoma"] = nombre_comunidad
    df_bloque_largo["nivel_territorial"] = "provincia"

    es_total_comunidad = (
        df_bloque_largo["territorio"] == total_comunidad
    )

    df_bloque_largo.loc[
        es_total_comunidad,
        "nivel_territorial"
    ] = "comunidad_autonoma"

    df_bloque_largo.loc[
        es_total_comunidad,
        "territorio"
    ] = nombre_comunidad

    bloques_largos.append(
        df_bloque_largo[
            [
                "anio",
                "comunidad_autonoma",
                "nivel_territorial",
                "territorio",
                "categoria_delito",
                "tipo_delito",
                "indicador",
                "valor"
            ]
        ]
    )

df_delitos_espana = pd.concat(
    bloques_largos,
    ignore_index=True
)

df_delitos_espana["valor"] = (
    df_delitos_espana["valor"].astype("Int64")
)

print("Dimensiones:", df_delitos_espana.shape)
print(
    "Comunidades autónomas:",
    df_delitos_espana["comunidad_autonoma"].nunique()
)

display(df_delitos_espana.head(10))
```

```python
df_total_nacional = df_original.iloc[
    20:88,
    [159, 160, 161, 162]
].copy()

df_total_nacional.columns = [
    "categoria_delito",
    "tipo_delito",
    "indicador",
    "valor"
]

df_total_nacional[
    ["categoria_delito", "tipo_delito"]
] = df_total_nacional[
    ["categoria_delito", "tipo_delito"]
].ffill()

df_total_nacional["anio"] = 2023
df_total_nacional["comunidad_autonoma"] = pd.NA
df_total_nacional["nivel_territorial"] = "nacional"
df_total_nacional["territorio"] = "España"

df_total_nacional["valor"] = (
    df_total_nacional["valor"].astype("Int64")
)

df_total_nacional = df_total_nacional[
    [
        "anio",
        "comunidad_autonoma",
        "nivel_territorial",
        "territorio",
        "categoria_delito",
        "tipo_delito",
        "indicador",
        "valor"
    ]
]

df_delitos_completo = pd.concat(
    [
        df_delitos_espana,
        df_total_nacional
    ],
    ignore_index=True
)

print("Dimensiones finales:", df_delitos_completo.shape)

display(
    df_delitos_completo["nivel_territorial"]
    .value_counts()
    .rename_axis("nivel_territorial")
    .reset_index(name="cantidad_filas")
)
```

```python
claves_nacionales = [
    "categoria_delito",
    "tipo_delito",
    "indicador"
]

suma_comunidades = (
    df_delitos_completo[
        df_delitos_completo["nivel_territorial"]
        == "comunidad_autonoma"
    ]
    .groupby(claves_nacionales, as_index=False)["valor"]
    .sum(min_count=1)
    .rename(columns={"valor": "suma_comunidades"})
)

totales_nacionales = (
    df_delitos_completo[
        df_delitos_completo["nivel_territorial"]
        == "nacional"
    ][claves_nacionales + ["valor"]]
    .rename(columns={"valor": "total_nacional"})
)

control_nacional = suma_comunidades.merge(
    totales_nacionales,
    on=claves_nacionales,
    how="outer",
    validate="one_to_one"
)

control_nacional["diferencia"] = (
    control_nacional["suma_comunidades"]
    - control_nacional["total_nacional"]
)

display(
    control_nacional["diferencia"]
    .value_counts(dropna=False)
    .rename_axis("diferencia")
    .reset_index(name="cantidad")
)
```

```python
claves_autonomicas = [
    "comunidad_autonoma",
    "categoria_delito",
    "tipo_delito",
    "indicador"
]

suma_provincias = (
    df_delitos_completo[
        df_delitos_completo["nivel_territorial"]
        == "provincia"
    ]
    .groupby(claves_autonomicas, as_index=False)["valor"]
    .sum(min_count=1)
    .rename(columns={"valor": "suma_provincias"})
)

totales_comunidades = (
    df_delitos_completo[
        df_delitos_completo["nivel_territorial"]
        == "comunidad_autonoma"
    ][claves_autonomicas + ["valor"]]
    .rename(columns={"valor": "total_comunidad"})
)

control_autonomico = suma_provincias.merge(
    totales_comunidades,
    on=claves_autonomicas,
    how="left",
    validate="one_to_one"
)

control_autonomico["diferencia"] = (
    control_autonomico["suma_provincias"]
    - control_autonomico["total_comunidad"]
)

print(
    "Comunidades comprobadas:",
    control_autonomico["comunidad_autonoma"].nunique()
)

display(
    control_autonomico["diferencia"]
    .value_counts(dropna=False)
    .rename_axis("diferencia")
    .reset_index(name="cantidad")
)
```

```python
claves_registro = [
    "anio",
    "nivel_territorial",
    "territorio",
    "categoria_delito",
    "tipo_delito",
    "indicador"
]

resumen_calidad = pd.DataFrame({
    "control": [
        "Filas totales",
        "Registros duplicados",
        "Valores nulos en valor",
        "Valores negativos",
        "Filas nacionales sin comunidad autónoma",
        "Categorías de delito",
        "Tipos de delito",
        "Indicadores"
    ],
    "resultado": [
        len(df_delitos_completo),
        df_delitos_completo.duplicated(
            subset=claves_registro
        ).sum(),
        df_delitos_completo["valor"].isna().sum(),
        df_delitos_completo["valor"].lt(0).sum(),
        (
            (
                df_delitos_completo["nivel_territorial"]
                == "nacional"
            )
            & (
                df_delitos_completo["comunidad_autonoma"]
                .isna()
            )
        ).sum(),
        df_delitos_completo["categoria_delito"].nunique(),
        df_delitos_completo["tipo_delito"].nunique(),
        df_delitos_completo["indicador"].nunique()
    ]
})

display(resumen_calidad)
```

```python
df_final = df_delitos_completo.copy()

columnas_texto = [
    "comunidad_autonoma",
    "nivel_territorial",
    "territorio",
    "categoria_delito",
    "tipo_delito",
    "indicador"
]

for columna in columnas_texto:
    df_final[columna] = (
        df_final[columna]
        .astype("string")
        .str.strip()
    )

df_final["anio"] = df_final["anio"].astype("Int64")
df_final["valor"] = df_final["valor"].astype("Int64")

nombre_csv = "delitos_informaticos_2023_limpio.csv"

df_final.to_csv(
    nombre_csv,
    index=False,
    encoding="utf-8",
    na_rep=""
)

print("Archivo creado:", nombre_csv)
print("Filas exportadas:", len(df_final))

files.download(nombre_csv)
```

```python
from google.cloud import bigquery

project_id = "cool-framing-412121"
dataset_id = "delitos_informaticos"

cliente_bigquery = bigquery.Client(
    project=project_id,
    location="EU"
)

dataset = cliente_bigquery.get_dataset(
    f"{project_id}.{dataset_id}"
)

print("Conexión con BigQuery correcta.")
print("Dataset:", dataset.full_dataset_id)
print("Ubicación:", dataset.location)
```

```python
table_id = (
    "cool-framing-412121."
    "delitos_informaticos."
    "delitos_informaticos_2023"
)

esquema = [
    bigquery.SchemaField("anio", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField(
        "comunidad_autonoma",
        "STRING",
        mode="NULLABLE"
    ),
    bigquery.SchemaField(
        "nivel_territorial",
        "STRING",
        mode="REQUIRED"
    ),
    bigquery.SchemaField(
        "territorio",
        "STRING",
        mode="REQUIRED"
    ),
    bigquery.SchemaField(
        "categoria_delito",
        "STRING",
        mode="REQUIRED"
    ),
    bigquery.SchemaField(
        "tipo_delito",
        "STRING",
        mode="REQUIRED"
    ),
    bigquery.SchemaField(
        "indicador",
        "STRING",
        mode="REQUIRED"
    ),
    bigquery.SchemaField(
        "valor",
        "INTEGER",
        mode="NULLABLE"
    )
]

configuracion_carga = bigquery.LoadJobConfig(
    schema=esquema,
    create_disposition=(
        bigquery.CreateDisposition.CREATE_IF_NEEDED
    ),
    write_disposition=(
        bigquery.WriteDisposition.WRITE_EMPTY
    )
)

trabajo_carga = cliente_bigquery.load_table_from_dataframe(
    df_final,
    table_id,
    job_config=configuracion_carga,
    location="EU"
)

trabajo_carga.result()

tabla = cliente_bigquery.get_table(table_id)

print("Tabla cargada correctamente.")
print("Tabla:", tabla.full_table_id)
print("Filas:", tabla.num_rows)
print("Columnas:", len(tabla.schema))
```

```python
consulta_validacion = """
SELECT
    nivel_territorial,
    COUNT(*) AS cantidad_filas,
    COUNT(valor) AS valores_informados,
    COUNTIF(valor IS NULL) AS valores_nulos
FROM
    `cool-framing-412121.delitos_informaticos.delitos_informaticos_2023`
GROUP BY
    nivel_territorial
ORDER BY
    CASE nivel_territorial
        WHEN 'provincia' THEN 1
        WHEN 'comunidad_autonoma' THEN 2
        WHEN 'nacional' THEN 3
    END
"""

configuracion_consulta = bigquery.QueryJobConfig(
    maximum_bytes_billed=10_485_760
)

resultado_validacion = cliente_bigquery.query(
    consulta_validacion,
    job_config=configuracion_consulta,
    location="EU"
).to_dataframe()

display(resultado_validacion)
```
